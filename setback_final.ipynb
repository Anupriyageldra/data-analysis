```python
import pandas as pd
import geopandas as gpd
from shapely import wkt
from shapely.geometry import LineString, Polygon, MultiPolygon
import math
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

# Load CSVs directly (update paths as needed)
parcels_path = "/content/parcels_abu_ftaira1.csv"
roads_path = "/content/abu_fatira road 2.csv"

parcels_df = pd.read_csv(parcels_path)
roads_df = pd.read_csv(roads_path)

# Determine geometry column in roads_df
def find_geometry_column(df):
    for col in df.columns:
        try:
            wkt.loads(df[col].iloc[0])
            return col
        except:
            continue
    raise ValueError("No valid WKT geometry column found in roads data")

roads_geom_col = find_geometry_column(roads_df)

parcels_df["geometry"] = parcels_df["geom"].apply(wkt.loads)
roads_df["geometry"] = roads_df[roads_geom_col].apply(wkt.loads)

parcels_gdf = gpd.GeoDataFrame(parcels_df, geometry="geometry", crs="EPSG:4326")
roads_gdf = gpd.GeoDataFrame(roads_df, geometry="geometry", crs="EPSG:4326")

# Project to metric CRS
parcels_gdf = parcels_gdf.to_crs(epsg=32636)
roads_gdf = roads_gdf.to_crs(parcels_gdf.crs)

# Get open frontages
def get_open_frontages(parcel_geom, obstacles_gdf, min_clearance=0.5):
    open_frontages = []
    if parcel_geom.is_empty:
        return open_frontages
    if isinstance(parcel_geom, MultiPolygon):
        parcel_geom = max(parcel_geom.geoms, key=lambda p: p.area)

    coords = list(parcel_geom.exterior.coords)
    obstacle_union = obstacles_gdf.geometry.unary_union

    for i in range(len(coords) - 1):
        side = LineString([coords[i], coords[i + 1]])
        midpoint = side.interpolate(0.5, normalized=True)

        if obstacle_union.distance(midpoint) > min_clearance:
            open_frontages.append(side)

    return open_frontages

# Helper to extract first coordinate from any geometry
from shapely.geometry.base import BaseGeometry

def extract_first_coord(geom: BaseGeometry):
    if geom.is_empty:
        return None
    if hasattr(geom, 'geoms'):
        for g in geom.geoms:
            pt = extract_first_coord(g)
            if pt:
                return pt
    elif hasattr(geom, 'coords'):
        coords = list(geom.coords)
        if coords:
            return coords[0]
    return None

# Refined rectangular setback generator
from shapely.ops import split

def get_extended_rectangle(front_line, obstacles, max_length=25):
    if front_line.geom_type == "MultiLineString":
        front_line = max(front_line.geoms, key=lambda g: g.length)
    if front_line.geom_type != "LineString" or len(front_line.coords) < 2:
        return None

    (x1, y1), (x2, y2) = front_line.coords[:2]
    dx, dy = x2 - x1, y2 - y1
    length = math.hypot(dx, dy)
    if length == 0:
        return None

    ux, uy = -dy / length, dx / length

    p1_start = (x1, y1)
    p2_start = (x2, y2)
    p1_ext = (x1 + ux * max_length, y1 + uy * max_length)
    p2_ext = (x2 + ux * max_length, y2 + uy * max_length)

    line1 = LineString([p1_start, p1_ext])
    line2 = LineString([p2_start, p2_ext])

    for obs in obstacles.geometry:
        if line1.intersects(obs):
            inter = line1.intersection(obs)
            new_p1 = extract_first_coord(inter)
            if new_p1:
                p1_ext = new_p1

        if line2.intersects(obs):
            inter = line2.intersection(obs)
            new_p2 = extract_first_coord(inter)
            if new_p2:
                p2_ext = new_p2

    polygon = Polygon([p1_start, p2_start, p2_ext, p1_ext])
    if polygon.is_valid:
        return polygon
    return None

# Store setback polygons
setback_results = []

for idx, parcel in parcels_gdf.iterrows():
    parcel_geom = parcel.geometry
    parcel_id = parcel['parcel_uid']

    obstacles = parcels_gdf[parcels_gdf.parcel_uid != parcel_id].copy()
    obstacles = pd.concat([obstacles, roads_gdf], ignore_index=True)

    frontages = get_open_frontages(parcel_geom, obstacles, min_clearance=0.5)
    print(f"Parcel {parcel_id} has {len(frontages)} open frontages.")

    for i, frontage in enumerate(frontages):
        setback_poly = get_extended_rectangle(frontage, obstacles, 25)
        if setback_poly and setback_poly.is_valid and setback_poly.area > 1:
            setback_results.append({
                "parcel_uid": parcel_id,
                "frontage_id": i,
                "area_sqm": setback_poly.area,
                "geometry": setback_poly
            })

# Convert to GeoDataFrame and save CSV
setback_gdf = gpd.GeoDataFrame(setback_results, geometry="geometry", crs=parcels_gdf.crs)

# Calculate total setback area per parcel
total_areas = setback_gdf.groupby("parcel_uid")["area_sqm"].sum().reset_index()
total_areas.rename(columns={"area_sqm": "total_setback_area_sqm"}, inplace=True)

# Merge total area into individual setbacks
setback_gdf = setback_gdf.merge(total_areas, on="parcel_uid", how="left")

# Save to CSV with WKT geometries
setback_gdf["geometry"] = setback_gdf["geometry"].apply(lambda geom: geom.wkt if geom else None)
setback_gdf.to_csv("setback_areas.csv", index=False)

# Merge total setback area into parcels_gdf
parcels_gdf = parcels_gdf.merge(total_areas, on="parcel_uid", how="left")
parcels_gdf["total_setback_area_sqm"].fillna(0, inplace=True)

# Visualize all setbacks
def visualize_all_setbacks():
    fig, ax = plt.subplots(figsize=(12, 12))
    parcels_gdf.plot(ax=ax, color='none', edgecolor='gray', linewidth=0.5)
    roads_gdf.plot(ax=ax, color='black', linewidth=1, alpha=0.5)
    gpd.GeoDataFrame(setback_results, geometry=gpd.GeoSeries.from_wkt([r["geometry"] for r in setback_results]), crs=parcels_gdf.crs).plot(ax=ax, color='lightgreen', edgecolor='green', alpha=0.6)
    plt.title("All Parcels with Setback Areas")
    plt.xlabel("Easting (meters)")
    plt.ylabel("Northing (meters)")
    plt.grid(False)
    plt.show()

# Visualize a parcel's setbacks
def visualize_setback_matplotlib(parcel_id):
    parcel_row = parcels_gdf[parcels_gdf["parcel_uid"] == parcel_id]
    if parcel_row.empty:
        print("❌ Parcel not found!")
        return

    fig, ax = plt.subplots(figsize=(10, 10))
    base = parcels_gdf.plot(ax=ax, color='none', edgecolor='gray', linewidth=0.5)
    parcel_row.plot(ax=base, color='lightblue', edgecolor='blue', linewidth=2)
    roads_gdf.plot(ax=base, color='black', linewidth=1, alpha=0.6)


    setbacks = setback_gdf[setback_gdf["parcel_uid"] == parcel_id]
    if not setbacks.empty:
        gpd.GeoDataFrame(setbacks, geometry=gpd.GeoSeries.from_wkt(setbacks["geometry"]), crs=parcels_gdf.crs).plot(ax=base, color='lightgreen', edgecolor='green', alpha=0.6)
    else:
        print("⚠️ No setback areas found for this parcel.")

    total_area = parcel_row["total_setback_area_sqm"].values[0]
    plt.title(f"Parcel ID: {parcel_id} | Total Setback Area: {total_area:.2f} sqm")
    plt.xlabel("Easting (meters)")
    plt.ylabel("Northing (meters)")
    plt.grid(False)
    plt.show()

# Interactive dropdown
parcel_dropdown = widgets.Dropdown(
    options=sorted(parcels_gdf["parcel_uid"].unique()),
    description='Parcel:',
    layout=widgets.Layout(width='50%')
)

button = widgets.Button(description="Visualize Setback", button_style='success')
all_button = widgets.Button(description="Visualize All Setbacks", button_style='info')
ui = widgets.VBox([parcel_dropdown, button, all_button])
display(ui)

button.on_click(lambda b: visualize_setback_matplotlib(parcel_dropdown.value))
all_button.on_click(lambda b: visualize_all_setbacks())

```
